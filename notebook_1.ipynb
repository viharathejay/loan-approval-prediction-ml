# Outlier detection

**seminar 2 page 10**

Defining a reusable python funtion which uses the quantile method to identify values which falling bellow q1- 1.5 IQR or above Q3+1.5x IQR as outliers

```python
#IQR funtion for finding outliers
def find_outliers_IQR(df):
      q1 = df.quantile(0.25)
      q3 = df.quantile(0.75)
      IQR = q3 - q1
      outliers = df[((df < (q1 - 1.5 * IQR)) | (df > (q3 + 1.5 * IQR)))]
      return outliers
```
